```python
import torch
import torchvision.models as models
from torchvision import transforms
from PIL import Image
```

```python
# Path to the saved VGG19 model weights
model_path = "C:\\Users\\ezeki\\OneDrive\\Desktop\\MTechProject\\best_vgg19_model.pth"

num_classes = 3  

# Load the VGG19 model
model = models.vgg19()
model.classifier = torch.nn.Sequential(
    torch.nn.Dropout(0.5),
    torch.nn.Linear(model.classifier[0].in_features, num_classes)
)
model = torch.load(model_path, map_location=torch.device('cpu'), weights_only=False)
 
model.eval()  # Set the model to evaluation mode
print("VGG19 model loaded successfully!")
```

```python
# Define preprocessing transformations
preprocess = transforms.Compose([
    transforms.Resize((224, 224)),  
    transforms.ToTensor(),        
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),  
])
```

```python
# Load and preprocess the image
image_path = "C:\\Users\\ezeki\\OneDrive\\Desktop\\MTechProject\\dataset\\Stroke_classification\\Ischemic\\Amina DWI-16.jpg_Ischemic_21.png"
image = Image.open(image_path).convert("RGB")  # Ensure the image is in RGB format

input_tensor = preprocess(image).unsqueeze(0)  # Add batch dimension
```

```python
# Perform inference
with torch.no_grad():
    output = model(input_tensor)  
    _, predicted_class = output.max(1)  # Get the class with the highest score

# Map the predicted class index to the class label
class_labels = ["Class1", "Ischemic", "Normal"]  # Update with your actual class names
predicted_label = class_labels[predicted_class.item()]

print(f"Predicted class: {predicted_label}")
```
